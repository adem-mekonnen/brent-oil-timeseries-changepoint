```python
import sys, os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
```

```python
# Force non-interactive backend to avoid 'get_data_path' or 'backend' errors
import matplotlib
matplotlib.use('Agg') 
```

```python
# Load Data
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
from src.data_loader import load_oil_data
df = load_oil_data()
```

```python
# 1. Trend Analysis (Moving Averages)
df['MA20'] = df['Price'].rolling(window=20).mean()
df['MA100'] = df['Price'].rolling(window=100).mean()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['Price'], label='Price', alpha=0.4)
plt.plot(df['Date'], df['MA100'], label='100-Day Trend', color='red')
plt.title("Brent Oil Price Trend Analysis")
plt.legend()
plt.savefig("trend_analysis.png") # Save for report
```

```python
# 2. Stationarity Testing (ADF Test)
# Critical Step: Proving the data needs a Change Point model
print("--- Stationarity Test (ADF) ---")
result = adfuller(df['Price'])
print(f'ADF Statistic: {result[0]}')
print(f'p-value: {result[1]}') 
# Note: If p-value > 0.05, it is non-stationary.
```

```python
# 3. Volatility Patterns (Log Returns)
plt.figure(figsize=(12, 4))
plt.plot(df['Date'], df['Log_Return'], color='orange', alpha=0.7)
plt.title("Daily Volatility Patterns (Log Returns)")
plt.savefig("volatility.png") # Save for report
```

### 1.4 Assumptions and Limitations
**Assumptions:**
- **Structural Integrity:** We assume that oil price changes occur as discrete "switch points" rather than gradual transitions.
- **Distribution:** We assume log returns are approximately normally distributed within each regime.

**Limitations (Correlation vs. Causation):**
- A major limitation of this analysis is the distinction between **statistical correlation** and **causality**. 
- The model identifies a "Structural Break" when the mean price shifts significantly. While a break might align perfectly with the date of an event (e.g., Feb 24, 2022), this proves a **temporal correlation**. 
- **Causality** is harder to prove because multiple factors (interest rates, USD index, and storage levels) often coincide with geopolitical events. Our analysis identifies the *impact* of the shift, but the *cause* is an inference based on historical context.
